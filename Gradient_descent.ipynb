```python
import matplotlib.pyplot as plt
```

```python
x=2
lr=0.1
X=[]
Y=[]
```

```python
for i in range(50):
    X.append(x)
    Y.append((x+3)**2)
    gradient=2*(x+3)
    x = x - lr * gradient
```

```python
print("Minimum x: ",x)
print("Minimum y:",(x+3)**2)
```

```python
plt.plot(X,Y)
plt.xlabel("x")

plt.show()
```
